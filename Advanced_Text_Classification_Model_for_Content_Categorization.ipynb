```python
import os
import requests

INPUT_FILE_PATH = "/content/drive/MyDrive/Dataset For Text Classification Model/seo_services_urls.txt"

OUTPUT_FILE_PATH = "/content/drive/MyDrive/Dataset For Text Classification Model/valid_urls.txt"

def read_urls_from_file(file_path):
    try:
      with open(file_path, 'r') as file:
        urls = [line.strip() for line in file if line.strip()]
        print(f"Total URLs read:{len(urls)}")
        return urls
    except FileNotFoundError:
      print(f"File not found: {file_path}")
      return[]

def is_valid_url(url):
  try:
    response = requests.head(url,timeout=5)
    return response.status_code == 200
  except requests.exceptions.RequestException:
    return False

def prepare_data(input_file, output_file):
    # Read all URLs from the input file
    urls = read_urls_from_file(input_file)
    if not urls:
        print("No URLs were found to process. Exiting.")
        return

    # Lists to keep track of valid and invalid URLs
    valid_urls = []
    invalid_urls = []

    # Validate each URL
    for url in urls:
        print(f"Checking URL: {url}")
        if is_valid_url(url):
            # If valid, add to the valid_urls list
            print(f"Valid URL: {url}")
            valid_urls.append(url)
        else:
            # If invalid, add to the invalid_urls list
            print(f"Invalid or inaccessible URL: {url}")
            invalid_urls.append(url)

    # Save valid URLs to the output file
    try:
        with open(output_file, 'w') as file:
            # Write each valid URL on a new line
            file.write("\n".join(valid_urls))
        print(f"\nValid URLs successfully saved to: {output_file}")
    except Exception as e:
        print(f"Error writing to the output file: {e}")

    # Print a summary of the results
    print("\nData Preparation Summary:")
    print(f"Total URLs processed: {len(urls)}")
    print(f"Valid URLs: {len(valid_urls)}")
    print(f"Invalid URLs: {len(invalid_urls)}")

    # Optionally display invalid URLs for further inspection
    if invalid_urls:
        print("\nInvalid URLs:")
        print("\n".join(invalid_urls))

# Step 4: Execute the validation process
if __name__ == "__main__":
    """
    Main execution block:
    - Specifies file paths for input and output.
    - Calls the prepare_data function to validate URLs and save valid ones.
    """
    prepare_data(INPUT_FILE_PATH, OUTPUT_FILE_PATH)
```

```python
import os
import requests
from bs4 import BeautifulSoup
import pandas as pd

# File path for valid URLs (input file)
INPUT_FILE_PATH = "/content/drive/MyDrive/Dataset For Text Classification Model/valid_urls.txt"

# File path for saving the scraped content
OUTPUT_CSV_PATH = "/content/drive/MyDrive/Dataset For Text Classification Model/scraped_content.csv"

# Step 1: Function to read valid URLs from a file
def read_valid_urls(file_path):

    try:
        with open(file_path, 'r') as file:
            urls = [line.strip() for line in file if line.strip()]  # Remove empty lines and whitespace
        print(f"Total valid URLs found: {len(urls)}")
        return urls
    except FileNotFoundError:
        print(f"Error: File not found at {file_path}")
        return []

# Step 2: Function to scrape content from a URL
def scrape_content(url):

    try:
        response = requests.get(url, timeout=10)  # Send an HTTP GET request with a 10-second timeout
        response.raise_for_status()  # Raise an error if the response status is not 200 (OK)
        soup = BeautifulSoup(response.text, 'html.parser')  # Parse the HTML content

        # Extract the title of the webpage
        title = soup.title.string.strip() if soup.title else "No Title Found"

        # Extract text content from <p> tags
        paragraphs = soup.find_all('p')  # Find all paragraph tags
        content = " ".join([p.get_text().strip() for p in paragraphs])  # Concatenate all paragraph texts

        return title, content
    except Exception as e:
        print(f"Error scraping {url}: {e}")
        return None, None

# Step 3: Function to save scraped data to a CSV file
def save_to_csv(data, output_file):

    try:
        df = pd.DataFrame(data, columns=["URL", "Title", "Content"])  # Create a DataFrame
        df.to_csv(output_file, index=False)  # Save the DataFrame to a CSV file
        print(f"Scraped content successfully saved to: {output_file}")
    except Exception as e:
        print(f"Error saving to CSV: {e}")

# Step 4: Main function to handle the scraping process
def scrape_urls(input_file, output_file):

    urls = read_valid_urls(input_file)  # Read URLs from the input file
    if not urls:
        print("No valid URLs found to process. Exiting.")
        return

    scraped_data = []  # List to store the scraped data

    # Scrape content from each URL
    for i, url in enumerate(urls):
        print(f"Scraping URL {i + 1}/{len(urls)}: {url}")
        title, content = scrape_content(url)  # Scrape the title and content

        if title and content:
            # Append the result to the scraped_data list
            scraped_data.append({"URL": url, "Title": title, "Content": content})

            # Show a preview of the content for the first 20 URLs (up to 500 characters)
            if i < 20:
                print(f"\nPreview of {url}:")
                print(f"Title: {title}")
                print(content[:500])  # Display the first 500 characters of the content
                print("\n" + "-" * 50 + "\n")

    # Save all scraped data to the output CSV file
    save_to_csv(scraped_data, output_file)

# Step 5: Run the scraping function
if __name__ == "__main__":

    scrape_urls(INPUT_FILE_PATH, OUTPUT_CSV_PATH)
```

```python
import os
import pandas as pd
import spacy
from spacy.lang.en.stop_words import STOP_WORDS

# Step 1: Initialize spaCy and load the English language model
def initialize_spacy():

    try:
        # Load the spaCy English model
        nlp = spacy.load("en_core_web_sm")
        print("SpaCy English model loaded successfully.")
        return nlp
    except Exception as e:
        print(f"Error loading SpaCy model: {e}")
        raise

# Step 2: Define a function for text cleaning
def clean_text_spacy(nlp, text):

    try:
        # Process the text using spaCy
        doc = nlp(text.lower())  # Convert to lowercase for uniformity

        # Extract only alphabetic words, excluding stopwords
        cleaned_words = [
            token.text for token in doc
            if token.is_alpha and token.text not in STOP_WORDS
        ]

        # Join the cleaned words into a single string
        return " ".join(cleaned_words)
    except Exception as e:
        print(f"Error cleaning text: {e}")
        return ""

# Step 3: Preprocess the dataset
def preprocess_dataset(input_file, output_file, nlp):

    try:
        # Load the input dataset into a pandas DataFrame
        df = pd.read_csv(input_file)
        print(f"Loaded data from {input_file}. Total rows: {len(df)}")
    except FileNotFoundError:
        print(f"Error: File not found at {input_file}")
        return
    except pd.errors.EmptyDataError:
        print(f"Error: No data found in {input_file}")
        return

    try:
        # Ensure the dataset contains a 'Content' column
        if 'Content' not in df.columns:
            raise KeyError("'Content' column not found in the dataset.")

        # Apply text cleaning to the 'Content' column
        print("\nProcessing text data...\n")
        df['Cleaned_Content'] = df['Content'].apply(lambda x: clean_text_spacy(nlp, str(x)))

        # Display a preview of the cleaned content
        print("Sample Processed Output:")
        print(df[['Content', 'Cleaned_Content']].head(40))  # Display the first 40 rows
    except Exception as e:
        print(f"Error during text processing: {e}")
        return

    try:
        # Save the cleaned dataset to a new file
        df.to_csv(output_file, index=False)
        print(f"\nProcessed content saved to: {output_file}")
    except Exception as e:
        print(f"Error saving processed dataset: {e}")

# Step 4: Main execution block
if __name__ == "__main__":
    # Initialize spaCy for text processing
    nlp = initialize_spacy()

    # File paths (adjust these paths as needed)
    INPUT_CSV_PATH = "/content/drive/MyDrive/Dataset For Text Classification Model/scraped_content.csv"
    OUTPUT_CSV_PATH = "/content/drive/MyDrive/Dataset For Text Classification Model/processed_content.csv"

    # Run the preprocessing function
    preprocess_dataset(INPUT_CSV_PATH, OUTPUT_CSV_PATH, nlp)
```

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.metrics.pairwise import cosine_similarity

def dynamic_categorization_with_clusters_and_categories(input_csv, output_csv, predefined_categories, num_categories=5):

    try:
        # Step 1: Load the cleaned dataset
        print("Loading the dataset...")
        df = pd.read_csv(input_csv)

        # Verify that the required column 'Cleaned_Content' exists
        if 'Cleaned_Content' not in df.columns:
            raise ValueError("The 'Cleaned_Content' column is missing in the dataset.")
        print(f"Dataset loaded successfully. Total rows: {len(df)}")

        # Step 2: Vectorize the content using TF-IDF
        print("Vectorizing text content using TF-IDF...")
        vectorizer = TfidfVectorizer(
            max_features=1000,  # Use only the top 1000 terms for efficiency
            stop_words='english'  # Remove common stopwords to focus on meaningful words
        )
        X = vectorizer.fit_transform(df['Cleaned_Content'])  # Convert text into numerical TF-IDF vectors
        print("Text content vectorized successfully.")

        # Step 3: Apply KMeans clustering to create clusters
        print(f"Clustering content into {num_categories} clusters using KMeans...")
        kmeans = KMeans(n_clusters=num_categories, random_state=42)  # Initialize KMeans with a fixed random state
        df['Cluster'] = kmeans.fit_predict(X)  # Assign each row to a cluster
        print("Clustering completed successfully.")

        # Step 4: Generate representative terms for each cluster
        print("Extracting representative terms for each cluster...")
        terms = vectorizer.get_feature_names_out()  # Get the feature names (words/phrases)
        cluster_centers = kmeans.cluster_centers_  # Access the cluster centers in the TF-IDF space
        cluster_labels = {}  # Store a descriptive label for each cluster

        # For each cluster, find the top terms
        for i in range(num_categories):
            # Sort terms by their importance to the cluster
            top_indices = cluster_centers[i].argsort()[-10:][::-1]
            top_terms = [terms[idx] for idx in top_indices]  # Get the top terms for this cluster
            cluster_labels[i] = " & ".join(top_terms[:3])  # Use the top 3 terms as the cluster label
        print(f"Cluster representative terms: {cluster_labels}")

        # Step 5: Map clusters to predefined categories
        print("Mapping clusters to predefined categories...")
        predefined_texts = list(predefined_categories.values())  # Get descriptions of predefined categories
        predefined_names = list(predefined_categories.keys())  # Get names of predefined categories
        predefined_vectors = vectorizer.transform(predefined_texts)  # Vectorize predefined category descriptions

        cluster_to_category_mapping = {}  # Map each cluster to its closest predefined category
        for cluster, label in cluster_labels.items():
            # Vectorize the representative terms for this cluster
            cluster_vector = vectorizer.transform([" ".join(label.split(" & "))])
            # Calculate similarity between this cluster and predefined categories
            similarities = cosine_similarity(cluster_vector, predefined_vectors)
            # Find the best matching predefined category
            best_match_index = similarities.argmax()
            best_category = predefined_names[best_match_index]
            cluster_to_category_mapping[cluster] = best_category  # Map the cluster to this category
            print(f"Cluster {cluster}: {label} -> {best_category}")

        # Assign user-friendly categories to the content based on their cluster
        df['Category'] = df['Cluster'].map(cluster_to_category_mapping)

        # Step 6: Save the categorized data to a CSV file
        print("Saving the categorized content with clusters and categories...")
        df.to_csv(output_csv, index=False)
        print(f"Categorized content saved to: {output_csv}")

        # Step 7: Display a preview of the first 40 rows
        print("\nPreview of the first 40 rows:")
        print(df[['Cleaned_Content', 'Cluster', 'Category']].head(40))

    except Exception as e:
        print(f"Error during content categorization: {e}")

# Example Usage
if __name__ == "__main__":
    # Path to input file containing cleaned content
    INPUT_CSV_PATH = "/content/drive/MyDrive/Dataset For Text Classification Model/processed_content.csv"
    # Path to save the categorized output
    OUTPUT_CSV_PATH = "/content/drive/MyDrive/Dataset For Text Classification Model/categorized_with_clusters.csv"

    # Define the predefined categories with their descriptions
    predefined_categories = {
        "SEO Services": "Search engine optimization, improving rankings, website traffic",
        "Content Writing": "Creating articles, blog posts, and SEO-friendly content",
        "Web Development": "Building websites, responsive design, and web performance",
        "Digital Marketing": "Online advertising, branding, social media, and marketing strategies",
        "AI and Machine Learning": "Artificial intelligence, machine learning, data-driven models"
    }

    # Run the dynamic categorization function
    dynamic_categorization_with_clusters_and_categories(INPUT_CSV_PATH, OUTPUT_CSV_PATH, predefined_categories, num_categories=5)
```

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
import random

def extract_refined_keywords_with_urls(input_csv, output_csv, n_unigrams=3, n_bigrams=5, n_trigrams=2):
    """
    Extract meaningful keywords (unigrams, bigrams, trigrams) for each URL and content dynamically.

    Args:
        input_csv (str): Path to the input CSV file with URLs and cleaned content.
        output_csv (str): Path to save the output CSV with refined keywords.
        n_unigrams (int): Number of single-word keywords to extract.
        n_bigrams (int): Number of two-word phrases to extract.
        n_trigrams (int): Number of three-word phrases to extract.

    Returns:
        None: Saves the results to a CSV and prints a preview for validation.
    """

    try:
        # Step 1: Load the dataset
        print("Loading dataset...")
        df = pd.read_csv(input_csv)

        # Verify that the required columns exist in the dataset
        required_columns = ['URL', 'Cleaned_Content']
        if not all(col in df.columns for col in required_columns):
            raise ValueError(f"The dataset must include the following columns: {required_columns}")
        print(f"Dataset loaded successfully. Total rows: {len(df)}")

        # Step 2: Initialize the TF-IDF Vectorizer
        print("Vectorizing text content using TF-IDF...")
        vectorizer = TfidfVectorizer(
            max_features=1000,  # Limit the features to the top 1000 terms for computational efficiency
            ngram_range=(1, 3),  # Extract unigrams, bigrams, and trigrams
            stop_words='english'  # Remove common English stop words to focus on meaningful terms
        )

        # Convert the 'Cleaned_Content' column into a TF-IDF matrix
        X = vectorizer.fit_transform(df['Cleaned_Content'])

        # Extract the feature names (i.e., the terms extracted by the TF-IDF vectorizer)
        feature_names = vectorizer.get_feature_names_out()
        print("TF-IDF vectorization complete.")

        # Step 3: Extract and refine keywords
        print("Extracting keywords for each row...")

        def refine_keywords(matrix, n_unigrams, n_bigrams, n_trigrams):
            """
            Extract and prioritize keywords for each row of content.

            Args:
                matrix (csr_matrix): TF-IDF matrix representing the content.
                n_unigrams (int): Number of single-word keywords to extract.
                n_bigrams (int): Number of two-word phrases to extract.
                n_trigrams (int): Number of three-word phrases to extract.

            Returns:
                list: A list of refined keywords for each row.
            """
            refined_keywords = []  # List to store refined keywords for each row

            for row in matrix.toarray():  # Convert sparse matrix to dense for processing
                # Sort the indices of terms by their importance (TF-IDF values) in descending order
                indices = row.argsort()[::-1]

                # Map sorted indices to the actual terms (keywords)
                top_keywords = [feature_names[i] for i in indices]

                # Separate terms into unigrams, bigrams, and trigrams
                unigrams = [kw for kw in top_keywords if len(kw.split()) == 1][:n_unigrams]
                bigrams = [kw for kw in top_keywords if len(kw.split()) == 2][:n_bigrams]
                trigrams = [kw for kw in top_keywords if len(kw.split()) == 3][:n_trigrams]

                # Combine all keywords while removing duplicates
                combined = list(dict.fromkeys(unigrams + bigrams + trigrams))

                # Shuffle the combined list for variety (optional, to avoid consistent order)
                random.shuffle(combined)

                # Append the refined keywords for this row
                refined_keywords.append(combined)

            return refined_keywords

        # Generate refined keywords for the dataset
        df['Refined_Keywords'] = refine_keywords(X, n_unigrams, n_bigrams, n_trigrams)

        # Step 4: Preview the result for validation
        print("\nPreview of Refined Keywords with URLs (First 40 Rows):")
        print(df[['URL', 'Cleaned_Content', 'Refined_Keywords']].head(40))

        # Step 5: Save the result to a CSV file
        print("Saving the results to the output file...")
        df.to_csv(output_csv, index=False)
        print(f"Refined keywords saved successfully to: {output_csv}")

    except Exception as e:
        # Handle and display any errors that occur during processing
        print(f"Error during keyword extraction: {e}")

# Example Usage
if __name__ == "__main__":
    # Define input and output file paths
    INPUT_CSV_PATH = "/content/drive/MyDrive/Dataset For Text Classification Model/processed_content.csv"
    OUTPUT_CSV_PATH = "/content/drive/MyDrive/Dataset For Text Classification Model/improved_refined_keywords_with_urls.csv"

    # Extract refined keywords for each URL and its content
    extract_refined_keywords_with_urls(INPUT_CSV_PATH, OUTPUT_CSV_PATH)
```

```python
import os
import requests

# File path for the input URLs
INPUT_FILE_PATH = "/content/drive/MyDrive/Dataset For Text Classification Model/seo_services_urls.txt"

# File path for saving valid URLs
OUTPUT_FILE_PATH = "/content/drive/MyDrive/Dataset For Text Classification Model/valid_urls.txt"

# Step 1: Function to read URLs from a file
def read_urls(file_path):
    """
    Reads URLs from the specified file.

    Use Case:
    This function reads URLs line by line from a text file. Each line should contain one URL.
    It ensures that the input file is processed correctly and avoids processing blank lines.

    Args:
        file_path (str): Path to the file containing URLs.

    Returns:
        list: A list of valid, non-empty URLs.
    """
    try:
        # Open the file in read mode and read all lines
        with open(file_path, 'r') as file:
            # Strip whitespace and filter out empty lines
            urls = [line.strip() for line in file if line.strip()]
        print(f"Total URLs read: {len(urls)}")
        return urls
    except FileNotFoundError:
        # If the file is missing, print an error message
        print(f"Error: File not found at {file_path}")
        return []

# Step 2: Function to validate a URL
def is_valid_url(url):
    """
    Validates if a URL is accessible.

    Use Case:
    This function checks if a URL exists and is accessible by sending a HEAD request.
    A HEAD request is faster than a GET request since it only retrieves headers without downloading the content.

    Args:
        url (str): The URL to validate.

    Returns:
        bool: True if the URL is accessible (status code 200), False otherwise.
    """
    try:
        # Send a HEAD request with a timeout of 5 seconds
        response = requests.head(url, timeout=5)
        # If the response status code is 200 (OK), the URL is valid
        return response.status_code == 200
    except requests.RequestException:
        # If there is a timeout or any other exception, the URL is invalid
        return False

# Step 3: Function to validate and save valid URLs
def prepare_data(input_file, output_file):
    """
    Validates URLs from an input file and saves valid ones to a new output file.

    Use Case:
    This function ties everything together:
    - Reads URLs from a file.
    - Validates each URL for accessibility.
    - Saves valid URLs to a separate file.
    This ensures we only process and store URLs that are active and accessible.

    Args:
        input_file (str): Path to the file containing input URLs.
        output_file (str): Path to save valid URLs.
    """
    # Read all URLs from the input file
    urls = read_urls(input_file)
    if not urls:
        print("No URLs were found to process. Exiting.")
        return

    # Lists to keep track of valid and invalid URLs
    valid_urls = []
    invalid_urls = []

    # Validate each URL
    for url in urls:
        print(f"Checking URL: {url}")
        if is_valid_url(url):
            # If valid, add to the valid_urls list
            print(f"Valid URL: {url}")
            valid_urls.append(url)
        else:
            # If invalid, add to the invalid_urls list
            print(f"Invalid or inaccessible URL: {url}")
            invalid_urls.append(url)

    # Save valid URLs to the output file
    try:
        with open(output_file, 'w') as file:
            # Write each valid URL on a new line
            file.write("\n".join(valid_urls))
        print(f"\nValid URLs successfully saved to: {output_file}")
    except Exception as e:
        print(f"Error writing to the output file: {e}")

    # Print a summary of the results
    print("\nData Preparation Summary:")
    print(f"Total URLs processed: {len(urls)}")
    print(f"Valid URLs: {len(valid_urls)}")
    print(f"Invalid URLs: {len(invalid_urls)}")

    # Optionally display invalid URLs for further inspection
    if invalid_urls:
        print("\nInvalid URLs:")
        print("\n".join(invalid_urls))

# Step 4: Execute the validation process
if __name__ == "__main__":
    """
    Main execution block:
    - Specifies file paths for input and output.
    - Calls the prepare_data function to validate URLs and save valid ones.
    """
    prepare_data(INPUT_FILE_PATH, OUTPUT_FILE_PATH)



import os
import requests
from bs4 import BeautifulSoup
import pandas as pd

# File path for valid URLs (input file)
INPUT_FILE_PATH = "/content/drive/MyDrive/Dataset For Text Classification Model/valid_urls.txt"

# File path for saving the scraped content
OUTPUT_CSV_PATH = "/content/drive/MyDrive/Dataset For Text Classification Model/scraped_content.csv"

# Step 1: Function to read valid URLs from a file
def read_valid_urls(file_path):
    """
    Reads valid URLs from the specified file.

    Use Case:
    This function ensures that only non-empty, properly formatted URLs are read from the file.
    It handles missing files gracefully by notifying the user.

    Args:
        file_path (str): Path to the file containing valid URLs.

    Returns:
        list: A list of valid URLs read from the file.
    """
    try:
        with open(file_path, 'r') as file:
            urls = [line.strip() for line in file if line.strip()]  # Remove empty lines and whitespace
        print(f"Total valid URLs found: {len(urls)}")
        return urls
    except FileNotFoundError:
        print(f"Error: File not found at {file_path}")
        return []

# Step 2: Function to scrape content from a URL
def scrape_content(url):
    """
    Scrapes the title and main content of a webpage.

    Use Case:
    This function retrieves the title and the main content (usually found in <p> tags) of the given URL.

    Args:
        url (str): The URL of the webpage to scrape.

    Returns:
        tuple: The title and the content of the webpage. Returns (None, None) if an error occurs.
    """
    try:
        response = requests.get(url, timeout=10)  # Send an HTTP GET request with a 10-second timeout
        response.raise_for_status()  # Raise an error if the response status is not 200 (OK)
        soup = BeautifulSoup(response.text, 'html.parser')  # Parse the HTML content

        # Extract the title of the webpage
        title = soup.title.string.strip() if soup.title else "No Title Found"

        # Extract text content from <p> tags
        paragraphs = soup.find_all('p')  # Find all paragraph tags
        content = " ".join([p.get_text().strip() for p in paragraphs])  # Concatenate all paragraph texts

        return title, content
    except Exception as e:
        print(f"Error scraping {url}: {e}")
        return None, None

# Step 3: Function to save scraped data to a CSV file
def save_to_csv(data, output_file):
    """
    Saves the scraped content to a CSV file.

    Use Case:
    This function organizes the scraped data into a structured format for easy analysis and storage.

    Args:
        data (list): A list of dictionaries containing URL, title, and content.
        output_file (str): Path to save the CSV file.

    Returns:
        None
    """
    try:
        df = pd.DataFrame(data, columns=["URL", "Title", "Content"])  # Create a DataFrame
        df.to_csv(output_file, index=False)  # Save the DataFrame to a CSV file
        print(f"Scraped content successfully saved to: {output_file}")
    except Exception as e:
        print(f"Error saving to CSV: {e}")

# Step 4: Main function to handle the scraping process
def scrape_urls(input_file, output_file):
    """
    Orchestrates the process of reading, scraping, and saving data.

    Use Case:
    This function ties everything together. It:
    - Reads valid URLs.
    - Scrapes content from each URL.
    - Displays a preview for the first 20 URLs.
    - Saves all scraped content to a CSV file.

    Args:
        input_file (str): Path to the file containing valid URLs.
        output_file (str): Path to save the scraped content.

    Returns:
        None
    """
    urls = read_valid_urls(input_file)  # Read URLs from the input file
    if not urls:
        print("No valid URLs found to process. Exiting.")
        return

    scraped_data = []  # List to store the scraped data

    # Scrape content from each URL
    for i, url in enumerate(urls):
        print(f"Scraping URL {i + 1}/{len(urls)}: {url}")
        title, content = scrape_content(url)  # Scrape the title and content

        if title and content:
            # Append the result to the scraped_data list
            scraped_data.append({"URL": url, "Title": title, "Content": content})

            # Show a preview of the content for the first 20 URLs (up to 500 characters)
            if i < 20:
                print(f"\nPreview of {url}:")
                print(f"Title: {title}")
                print(content[:500])  # Display the first 500 characters of the content
                print("\n" + "-" * 50 + "\n")

    # Save all scraped data to the output CSV file
    save_to_csv(scraped_data, output_file)

# Step 5: Run the scraping function
if __name__ == "__main__":
    """
    Entry point for the script:
    - Reads valid URLs from the input file.
    - Scrapes content and saves it to an output CSV file.
    """
    scrape_urls(INPUT_FILE_PATH, OUTPUT_CSV_PATH)


import os
import pandas as pd
import spacy
from spacy.lang.en.stop_words import STOP_WORDS

# Step 1: Initialize spaCy and load the English language model
def initialize_spacy():
    """
    Initializes spaCy's English language model.

    Why:
        This step prepares spaCy's natural language processing tools,
        which are essential for processing text (e.g., tokenizing, lemmatizing, removing stopwords).
    """
    try:
        # Load the spaCy English model
        nlp = spacy.load("en_core_web_sm")
        print("SpaCy English model loaded successfully.")
        return nlp
    except Exception as e:
        print(f"Error loading SpaCy model: {e}")
        raise

# Step 2: Define a function for text cleaning
def clean_text_spacy(nlp, text):
    """
    Cleans and processes text using spaCy.

    Why:
        - Converts text to lowercase for uniformity.
        - Removes digits, special characters, and symbols to keep meaningful words.
        - Removes common words (stopwords) that don't contribute to the meaning of the text.
        - Keeps only alphabetic words for clarity.

    Args:
        nlp: The spaCy language model for text processing.
        text: The input text to be cleaned.

    Returns:
        The cleaned text as a string.
    """
    try:
        # Process the text using spaCy
        doc = nlp(text.lower())  # Convert to lowercase for uniformity

        # Extract only alphabetic words, excluding stopwords
        cleaned_words = [
            token.text for token in doc
            if token.is_alpha and token.text not in STOP_WORDS
        ]

        # Join the cleaned words into a single string
        return " ".join(cleaned_words)
    except Exception as e:
        print(f"Error cleaning text: {e}")
        return ""

# Step 3: Preprocess the dataset
def preprocess_dataset(input_file, output_file, nlp):
    """
    Cleans the 'Content' column of a dataset and saves the cleaned data.

    Why:
        - Cleaning raw text ensures that the data is structured and useful for downstream analysis.
        - Saving the processed data allows easy reuse without repeating the cleaning steps.

    Args:
        input_file: Path to the CSV file containing raw text data.
        output_file: Path to save the cleaned dataset.
        nlp: The spaCy language model for text processing.
    """
    try:
        # Load the input dataset into a pandas DataFrame
        df = pd.read_csv(input_file)
        print(f"Loaded data from {input_file}. Total rows: {len(df)}")
    except FileNotFoundError:
        print(f"Error: File not found at {input_file}")
        return
    except pd.errors.EmptyDataError:
        print(f"Error: No data found in {input_file}")
        return

    try:
        # Ensure the dataset contains a 'Content' column
        if 'Content' not in df.columns:
            raise KeyError("'Content' column not found in the dataset.")

        # Apply text cleaning to the 'Content' column
        print("\nProcessing text data...\n")
        df['Cleaned_Content'] = df['Content'].apply(lambda x: clean_text_spacy(nlp, str(x)))

        # Display a preview of the cleaned content
        print("Sample Processed Output:")
        print(df[['Content', 'Cleaned_Content']].head(40))  # Display the first 40 rows
    except Exception as e:
        print(f"Error during text processing: {e}")
        return

    try:
        # Save the cleaned dataset to a new file
        df.to_csv(output_file, index=False)
        print(f"\nProcessed content saved to: {output_file}")
    except Exception as e:
        print(f"Error saving processed dataset: {e}")

# Step 4: Main execution block
if __name__ == "__main__":
    # Initialize spaCy for text processing
    nlp = initialize_spacy()

    # File paths (adjust these paths as needed)
    INPUT_CSV_PATH = "/content/drive/MyDrive/Dataset For Text Classification Model/scraped_content.csv"
    OUTPUT_CSV_PATH = "/content/drive/MyDrive/Dataset For Text Classification Model/processed_content.csv"

    # Run the preprocessing function
    preprocess_dataset(INPUT_CSV_PATH, OUTPUT_CSV_PATH, nlp)



import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.metrics.pairwise import cosine_similarity

def dynamic_categorization_with_clusters_and_categories(input_csv, output_csv, predefined_categories, num_categories=5):
    """
    Dynamically categorize text content into clusters, generate semantic labels for each cluster,
    and map them to predefined categories using KMeans clustering and cosine similarity.

    Args:
        input_csv (str): Path to the input CSV containing cleaned text data.
        output_csv (str): Path to save the categorized content with cluster and category information.
        predefined_categories (dict): Dictionary with predefined categories and their descriptions.
        num_categories (int): Number of clusters to generate using KMeans.

    Returns:
        None: Saves the categorized content to the output CSV file.
    """
    try:
        # Step 1: Load the cleaned dataset
        print("Loading the dataset...")
        df = pd.read_csv(input_csv)

        # Verify that the required column 'Cleaned_Content' exists
        if 'Cleaned_Content' not in df.columns:
            raise ValueError("The 'Cleaned_Content' column is missing in the dataset.")
        print(f"Dataset loaded successfully. Total rows: {len(df)}")

        # Step 2: Vectorize the content using TF-IDF
        print("Vectorizing text content using TF-IDF...")
        vectorizer = TfidfVectorizer(
            max_features=1000,  # Use only the top 1000 terms for efficiency
            stop_words='english'  # Remove common stopwords to focus on meaningful words
        )
        X = vectorizer.fit_transform(df['Cleaned_Content'])  # Convert text into numerical TF-IDF vectors
        print("Text content vectorized successfully.")

        # Step 3: Apply KMeans clustering to create clusters
        print(f"Clustering content into {num_categories} clusters using KMeans...")
        kmeans = KMeans(n_clusters=num_categories, random_state=42)  # Initialize KMeans with a fixed random state
        df['Cluster'] = kmeans.fit_predict(X)  # Assign each row to a cluster
        print("Clustering completed successfully.")

        # Step 4: Generate representative terms for each cluster
        print("Extracting representative terms for each cluster...")
        terms = vectorizer.get_feature_names_out()  # Get the feature names (words/phrases)
        cluster_centers = kmeans.cluster_centers_  # Access the cluster centers in the TF-IDF space
        cluster_labels = {}  # Store a descriptive label for each cluster

        # For each cluster, find the top terms
        for i in range(num_categories):
            # Sort terms by their importance to the cluster
            top_indices = cluster_centers[i].argsort()[-10:][::-1]
            top_terms = [terms[idx] for idx in top_indices]  # Get the top terms for this cluster
            cluster_labels[i] = " & ".join(top_terms[:3])  # Use the top 3 terms as the cluster label
        print(f"Cluster representative terms: {cluster_labels}")

        # Step 5: Map clusters to predefined categories
        print("Mapping clusters to predefined categories...")
        predefined_texts = list(predefined_categories.values())  # Get descriptions of predefined categories
        predefined_names = list(predefined_categories.keys())  # Get names of predefined categories
        predefined_vectors = vectorizer.transform(predefined_texts)  # Vectorize predefined category descriptions

        cluster_to_category_mapping = {}  # Map each cluster to its closest predefined category
        for cluster, label in cluster_labels.items():
            # Vectorize the representative terms for this cluster
            cluster_vector = vectorizer.transform([" ".join(label.split(" & "))])
            # Calculate similarity between this cluster and predefined categories
            similarities = cosine_similarity(cluster_vector, predefined_vectors)
            # Find the best matching predefined category
            best_match_index = similarities.argmax()
            best_category = predefined_names[best_match_index]
            cluster_to_category_mapping[cluster] = best_category  # Map the cluster to this category
            print(f"Cluster {cluster}: {label} -> {best_category}")

        # Assign user-friendly categories to the content based on their cluster
        df['Category'] = df['Cluster'].map(cluster_to_category_mapping)

        # Step 6: Save the categorized data to a CSV file
        print("Saving the categorized content with clusters and categories...")
        df.to_csv(output_csv, index=False)
        print(f"Categorized content saved to: {output_csv}")

        # Step 7: Display a preview of the first 40 rows
        print("\nPreview of the first 40 rows:")
        print(df[['Cleaned_Content', 'Cluster', 'Category']].head(40))

    except Exception as e:
        print(f"Error during content categorization: {e}")

# Example Usage
if __name__ == "__main__":
    # Path to input file containing cleaned content
    INPUT_CSV_PATH = "/content/drive/MyDrive/Dataset For Text Classification Model/processed_content.csv"
    # Path to save the categorized output
    OUTPUT_CSV_PATH = "/content/drive/MyDrive/Dataset For Text Classification Model/categorized_with_clusters.csv"

    # Define the predefined categories with their descriptions
    predefined_categories = {
        "SEO Services": "Search engine optimization, improving rankings, website traffic",
        "Content Writing": "Creating articles, blog posts, and SEO-friendly content",
        "Web Development": "Building websites, responsive design, and web performance",
        "Digital Marketing": "Online advertising, branding, social media, and marketing strategies",
        "AI and Machine Learning": "Artificial intelligence, machine learning, data-driven models"
    }

    # Run the dynamic categorization function
    dynamic_categorization_with_clusters_and_categories(INPUT_CSV_PATH, OUTPUT_CSV_PATH, predefined_categories, num_categories=5)



import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
import random

def extract_refined_keywords_with_urls(input_csv, output_csv, n_unigrams=3, n_bigrams=5, n_trigrams=2):
    """
    Extract meaningful keywords (unigrams, bigrams, trigrams) for each URL and content dynamically.

    Args:
        input_csv (str): Path to the input CSV file with URLs and cleaned content.
        output_csv (str): Path to save the output CSV with refined keywords.
        n_unigrams (int): Number of single-word keywords to extract.
        n_bigrams (int): Number of two-word phrases to extract.
        n_trigrams (int): Number of three-word phrases to extract.

    Returns:
        None: Saves the results to a CSV and prints a preview for validation.
    """

    try:
        # Step 1: Load the dataset
        print("Loading dataset...")
        df = pd.read_csv(input_csv)

        # Verify that the required columns exist in the dataset
        required_columns = ['URL', 'Cleaned_Content']
        if not all(col in df.columns for col in required_columns):
            raise ValueError(f"The dataset must include the following columns: {required_columns}")
        print(f"Dataset loaded successfully. Total rows: {len(df)}")

        # Step 2: Initialize the TF-IDF Vectorizer
        print("Vectorizing text content using TF-IDF...")
        vectorizer = TfidfVectorizer(
            max_features=1000,  # Limit the features to the top 1000 terms for computational efficiency
            ngram_range=(1, 3),  # Extract unigrams, bigrams, and trigrams
            stop_words='english'  # Remove common English stop words to focus on meaningful terms
        )

        # Convert the 'Cleaned_Content' column into a TF-IDF matrix
        X = vectorizer.fit_transform(df['Cleaned_Content'])

        # Extract the feature names (i.e., the terms extracted by the TF-IDF vectorizer)
        feature_names = vectorizer.get_feature_names_out()
        print("TF-IDF vectorization complete.")

        # Step 3: Extract and refine keywords
        print("Extracting keywords for each row...")

        def refine_keywords(matrix, n_unigrams, n_bigrams, n_trigrams):
            """
            Extract and prioritize keywords for each row of content.

            Args:
                matrix (csr_matrix): TF-IDF matrix representing the content.
                n_unigrams (int): Number of single-word keywords to extract.
                n_bigrams (int): Number of two-word phrases to extract.
                n_trigrams (int): Number of three-word phrases to extract.

            Returns:
                list: A list of refined keywords for each row.
            """
            refined_keywords = []  # List to store refined keywords for each row

            for row in matrix.toarray():  # Convert sparse matrix to dense for processing
                # Sort the indices of terms by their importance (TF-IDF values) in descending order
                indices = row.argsort()[::-1]

                # Map sorted indices to the actual terms (keywords)
                top_keywords = [feature_names[i] for i in indices]

                # Separate terms into unigrams, bigrams, and trigrams
                unigrams = [kw for kw in top_keywords if len(kw.split()) == 1][:n_unigrams]
                bigrams = [kw for kw in top_keywords if len(kw.split()) == 2][:n_bigrams]
                trigrams = [kw for kw in top_keywords if len(kw.split()) == 3][:n_trigrams]

                # Combine all keywords while removing duplicates
                combined = list(dict.fromkeys(unigrams + bigrams + trigrams))

                # Shuffle the combined list for variety (optional, to avoid consistent order)
                random.shuffle(combined)

                # Append the refined keywords for this row
                refined_keywords.append(combined)

            return refined_keywords

        # Generate refined keywords for the dataset
        df['Refined_Keywords'] = refine_keywords(X, n_unigrams, n_bigrams, n_trigrams)

        # Step 4: Preview the result for validation
        print("\nPreview of Refined Keywords with URLs (First 40 Rows):")
        print(df[['URL', 'Cleaned_Content', 'Refined_Keywords']].head(40))

        # Step 5: Save the result to a CSV file
        print("Saving the results to the output file...")
        df.to_csv(output_csv, index=False)
        print(f"Refined keywords saved successfully to: {output_csv}")

    except Exception as e:
        # Handle and display any errors that occur during processing
        print(f"Error during keyword extraction: {e}")

# Example Usage
if __name__ == "__main__":
    # Define input and output file paths
    INPUT_CSV_PATH = "/content/drive/MyDrive/Dataset For Text Classification Model/processed_content.csv"
    OUTPUT_CSV_PATH = "/content/drive/MyDrive/Dataset For Text Classification Model/improved_refined_keywords_with_urls.csv"

    # Extract refined keywords for each URL and its content
    extract_refined_keywords_with_urls(INPUT_CSV_PATH, OUTPUT_CSV_PATH)
```

```python
import ast
from collections import Counter

# Convert the string representation of lists in 'Refined_Keywords' to actual lists
try:
    df_keywords['Refined_Keywords_List'] = df_keywords['Refined_Keywords'].apply(ast.literal_eval)
except Exception as e:
    print(f"Error converting 'Refined_Keywords' column: {e}")
    # Fallback if conversion fails, e.g., if some entries are not valid list strings
    df_keywords['Refined_Keywords_List'] = df_keywords['Refined_Keywords'].apply(lambda x: ast.literal_eval(x) if isinstance(x, str) else [])

# Flatten the list of lists into a single list of all keywords
all_keywords = [keyword for sublist in df_keywords['Refined_Keywords_List'] for keyword in sublist]

# Count the frequency of each keyword
keyword_counts = Counter(all_keywords)

# Get the top N keywords (e.g., top 20)
top_n_keywords = keyword_counts.most_common(20)

print("Top 20 keywords and their counts:")
for keyword, count in top_n_keywords:
    print(f"- {keyword}: {count}")
```
